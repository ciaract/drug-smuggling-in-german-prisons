#### Create new column "Spice_NPS_g"
- The columns “Spice_g” and “NPS_Mischung_g” capture the same information in the dataset, but for different time periods—presumably due to a renaming of the data category (‘Spice_g’ through 2023, “NPS_Mischung_g” starting in 2024). Since the two columns never overlap (no prison year has a value in both at the same time), they can be merged into a continuous time series without any loss of information. 

```python
df["NPS_Mischung_g"] = df["Spice_g"].combine_first(df["NPS_Mischung_g"])
df;
```

#### Select only specific columns 

```python
df_NPS = df[["Jahr", "JVA", "Bundesland", "NPS_Papier_Konsumeinheiten_stck", "NPS_Reinsubstanz_ml", "NPS_Mischung_g", "NPS_cm2", ]]
df_NPS;
```

## Focus: Hamburg

#### Filter by state Baden-Württemberg 

```python
df_NPS_HH = df_NPS[df_NPS['Bundesland'] == 'Hamburg']
df_NPS_HH;
```

#### Group by state (not by individual correctional facilities) 

```python
df_NPS_HH_grouped = (
    df_NPS_HH.groupby("Jahr")[["NPS_Mischung_g", "NPS_Papier_Konsumeinheiten_stck"]]
    .sum()
    .reset_index()
)
df_NPS_HH_grouped
```

#### Adding trend line for the devolopment of NPS_Mischung_g" 

```python
import numpy as np
x = df_NPS_HH_grouped["Jahr"]
y = df_NPS_HH_grouped["NPS_Mischung_g"]
steigung, achsenabschnitt = np.polyfit(x, y, 1)
df_NPS_HH_grouped["Trend_Mischung_g"] = steigung * x + achsenabschnitt
df_NPS_HH_grouped
```

#### Calculate Projection for the Year 2026

```python
faktor = 365 / 190
maske = df_NPS_HH_grouped["Jahr"] == 2026

df_NPS_HH_grouped.loc[maske, "NPS_Mischung_g"] *= faktor
df_NPS_HH_grouped.loc[maske, "NPS_Papier_Konsumeinheiten_stck"] *= faktor
df_NPS_HH_grouped
```

#### Export dataframe

```python
df_NPS_HH_grouped.to_excel("results/df_NPS_HH_grouped_trend.xlsx", index=False)
```
